### Another neat little tool for developers: whatmodel
Ever looked at a model name like `Qwen3-30B-A3B-Instruct-GGUF:Q4_K_M` and wondered what half of it means?
whatmodel explains LLM model names in plain English. Give it an Ollama tag, a Hugging Face repo or a GGUF file name, and it breaks it down: family, size, tuning, compression, format.
```
> pip install whatmodel 
> whatmodel qwen3.5:4b
```

A few things I like about it:
Works offline. It reads the name itself, no API key needed.
It can also look up what the name doesn’t say (context length, license, tool support) from your local Ollama or Hugging Face.

It never guesses. Tokens it doesn’t know are marked “unknown”.
Every meaning lives in one YAML file, so adding a new token takes no Python.

````python
# uv tool install whatmodel or pip install whatmodel
import whatmodel

parsed = whatmodel.parse("llama3.1:8b-instruct-q4_K_M") 
print(f"Family: {parsed.family}")
print(f"Version: {parsed.version}")
print(f"Size: {parsed.size}")
print(f"Parameters: {parsed.parameters}")
print(f"Unknown: {parsed.unknown}")
````

````python
info = whatmodel.explain("qwen3.5:4b")  # the name plus lookups
info.family, info.size  # what the name says: ('qwen', '4B')
info.context_length, info.supports_tools  # looked up: (262144, True)
info.quantization, info.parameter_count  # looked up if possible, else from the name
print(info)  # plain text, one line per field
info.to_dict()  # the same data --json prints
````
